```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\lenovo\Desktop\raw\Flights_2018_2.csv")
```

```python
#convert date column into datetime 
df['FlightDate'] = pd.to_datetime(df['FlightDate'])
```

```python
#set date as the index
df = df.set_index('FlightDate')
```

```python
daily_flights = df.resample('D').size()
daily_flights.head()
```

```python
# calculate daily average departure delay 
daily_delay = df['DepDelay'].resample('D').mean()

daily_delay.head()
```

```python
# calculate weekly flight count
weekly_flights  = df.resample('W').size()

weekly_flights
```

```python
import matplotlib.pyplot as plt

daily_flights.plot(figsize=(12,5))

plt.title('Daily Flight Count - February 2018')
plt.xlabel('Date')
plt.ylabel('Number of Flights')
plt.show()
```
