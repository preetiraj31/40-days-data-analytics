## Group By

```python
# count flight by airline (This tells you how many flights each airline operates in the dataset.)
df.groupby('Marketing_Airline_Network').size()
```

```python
# Average departure delay by airline(This calculates the average departure delay for each airline.)
df.groupby('Marketing_Airline_Network')['DepDelay'].mean()
```

```python
# Multiple Calculation
df.groupby('Marketing_Airline_Network').agg({
    'DepDelay': 'mean',
    'ArrDelay': 'mean',
    'Distance': 'mean'
})
```

## Merge 


```python
# first create two separete DataFrames
airline_flights = df.groupby(
    'Marketing_Airline_Network'
).size().reset_index(name='Total_Flights')
```

```python
airline_delays = df.groupby(
    'Marketing_Airline_Network'
)['DepDelay'].mean().reset_index(name='Average_Delay')
```

```python
# combine them 
airline_summary = pd.merge(
    airline_flights,
    airline_delays,
    on='Marketing_Airline_Network',
    how='inner'
)
airline_summary.head()
```
