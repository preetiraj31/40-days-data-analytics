```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv(r"D:\archive\StudentsPerformance.csv")
```

```python
for c in ["math score", "reading score", "writing score"]:
    df[c] = pd.to_numeric(df[c])

    print(df.shape)
    print(df.isnull().sum())
    df.head()
```

## t-test(compare 2 groups)

```python
# Test 1: Do students who completed test prep score differently in math?

# H0: mean math score is the same for both groups
# H1: the means differ
done = df[df["test preparation course"] == "completed"]["math score"]
none = df[df["test preparation course"] == "none"]["math score"]

t, p = stats.ttest_ind(done, none, equal_var=False)
print(f"t = {t:.3f}, p = {p:.5f}")
print(done.mean(), none.mean())
```

```python
# Do males and females differ in math score?
m = df[df["gender"] == "male"]["math score"]
f = df[df["gender"] == "female"]["math score"]
print(stats.ttest_ind(m, f, equal_var=False))
```

```python
# Does lunch type affect math score?
std = df[df["lunch"] == "standard"]["math score"]
free = df[df["lunch"] == "free/reduced"]["math score"]
print(stats.ttest_ind(std, free, equal_var=False))
```

```python
print(stats.ttest_rel(df["reading score"], df["writing score"]))
```

## Chi-square tests (2 categorical variables)

```python
table = pd.crosstab(df["gender"], df["test preparation course"])
chi2, p, dof, expected = stats.chi2_contingency(table)
print(table)
print(f"chi2 = {chi2:.3f}, p = {p:.4f}, dof = {dof}")
print(expected) 
```

```python
# Is lunch type related to parental education?

table = pd.crosstab(df["lunch"], df["parental level of education"])
chi2, p, dof, expected = stats.chi2_contingency(table)
print(f"chi2 = {chi2:.3f}, p = {p:.4f}, dof = {dof}")
```

```python
# Is race/ethnicity group related to test prep?
table = pd.crosstab(df["race/ethnicity"], df["test preparation course"])
print(stats.chi2_contingency(table)[:3])
```

## ANOVA (3+ group)

```python
groups = [g["math score"].values for _, g in df.groupby("race/ethnicity")]
F, p = stats.f_oneway(*groups)
print(f"F = {F:.3f}, p = {p:.5f}")
```

```python
#reading score by parental education.
groups = [g["reading score"].values for _, g in df.groupby("parental level of education")]
print(stats.f_oneway(*groups))
```
